## Calculate returns

### Subtask:
Calculate the daily returns for each asset in the `data_cleaned` DataFrame.


```python
# Select the 'Close' price data
close_prices = data_cleaned['Close']

# Calculate the daily percentage change
daily_returns = close_prices.pct_change()

# Display the first few rows of the daily_returns DataFrame
display(daily_returns.head())
```

## Calculate portfolio returns

### Subtask:
Calculate the daily returns for a hypothetical portfolio based on the individual asset returns. This will require defining portfolio weights.


```python
# Define portfolio weights (equal weights in this example)
num_assets = daily_returns.shape[1]
weights = [1.0 / num_assets] * num_assets
portfolio_weights = pd.Series(weights, index=daily_returns.columns)

# Calculate the weighted daily returns
weighted_daily_returns = daily_returns * portfolio_weights

# Calculate the overall daily portfolio return
portfolio_daily_return = weighted_daily_returns.sum(axis=1)

# Display the first few rows of the portfolio daily return
display(portfolio_daily_return.head())

```

## Calculate wealth index

### Subtask:
Compute the wealth index over time based on the portfolio returns, assuming an initial investment.


```python
# Calculate the cumulative return
cumulative_return = (1 + portfolio_daily_return).cumprod()

# Assume an initial investment
initial_investment = 1000

# Calculate the wealth index
wealth_index = initial_investment * cumulative_return

# Display the first few values of the wealth index
display(wealth_index.head())
```

## Calculate performance metrics

### Subtask:
Calculate annualized return, annualized volatility, and maximum drawdown for the portfolio.


```python
import numpy as np

# Calculate annualized return
annualized_return = portfolio_daily_return.mean() * 252

# Calculate annualized volatility
annualized_volatility = portfolio_daily_return.std() * np.sqrt(252)

# Calculate maximum drawdown
# Calculate the peak value
peak = wealth_index.cummax()
# Calculate the drawdown
drawdown = (wealth_index - peak) / peak
# Calculate the maximum drawdown
max_drawdown = drawdown.min()

# Display the calculated metrics
print(f"Annualized Return: {annualized_return:.4f}")
print(f"Annualized Volatility: {annualized_volatility:.4f}")
print(f"Maximum Drawdown: {max_drawdown:.4f}")
```

## Define portfolio functions

### Subtask:
Define functions to calculate portfolio return and volatility given a set of weights.


```python
# Define the number of trading days in a year
trading_days = 252

def portfolio_return(weights, expected_returns):
    """
    Calculates the annualized portfolio return.

    Args:
        weights (np.ndarray): Array of portfolio weights.
        expected_returns (pd.Series): Daily expected returns for each asset.

    Returns:
        float: Annualized portfolio return.
    """
    return np.dot(weights.T, expected_returns) * trading_days

def portfolio_volatility(weights, annualized_cov_matrix):
    """
    Calculates the portfolio volatility.

    Args:
        weights (np.ndarray): Array of portfolio weights.
        annualized_cov_matrix (pd.DataFrame): Annualized covariance matrix.

    Returns:
        float: Portfolio volatility.
    """
    return np.sqrt(np.dot(weights.T, np.dot(annualized_cov_matrix, weights)))

print("Portfolio return and volatility functions defined.")
```

## Find gmv portfolio

### Subtask:
Use an optimizer to find the weights for the Global Minimum Volatility (GMV) portfolio.


```python
from scipy.optimize import minimize

# Annualize the covariance matrix
annualized_cov_matrix = cov_matrix * trading_days

# Define the objective function to minimize (portfolio volatility)
def objective_function(weights):
    return portfolio_volatility(weights, annualized_cov_matrix)

# Define the constraints
constraints = ({'type': 'eq', 'fun': lambda weights: np.sum(weights) - 1})
bounds = tuple((0, 1) for asset in range(len(daily_returns.columns))) # Weights must be between 0 and 1

# Define the initial guess (equal weights)
initial_weights = np.array([1./len(daily_returns.columns)] * len(daily_returns.columns))

# Run the optimizer
optimization_result = minimize(objective_function, initial_weights, method='SLSQP', bounds=bounds, constraints=constraints)

# Extract the optimal weights for the GMV portfolio
gmv_weights = optimization_result.x

# Display the GMV weights
print("Global Minimum Volatility Portfolio Weights:")
print(pd.Series(gmv_weights, index=daily_returns.columns))
```

## Find msr portfolio

### Subtask:
Use an optimizer to find the weights for the Maximum Sharpe Ratio (MSR) portfolio (this will require estimating expected returns).


**Reasoning**:
Calculate the annualized expected returns, define the negative Sharpe Ratio function, set up constraints and initial weights, run the optimizer to find MSR weights, and then display the results.



```python
# Calculate the annualized expected returns for each asset
expected_returns = daily_returns.mean() * trading_days

# Define the negative Sharpe Ratio function to minimize
def negative_sharpe_ratio(weights, expected_returns, annualized_cov_matrix):
    """
    Calculates the negative Sharpe Ratio of a portfolio.

    Args:
        weights (np.ndarray): Array of portfolio weights.
        expected_returns (pd.Series): Annualized expected returns for each asset.
        annualized_cov_matrix (pd.DataFrame): Annualized covariance matrix.

    Returns:
        float: Negative Sharpe Ratio.
    """
    port_return = portfolio_return(weights, expected_returns / trading_days) # Use daily expected returns for portfolio_return function
    port_volatility = portfolio_volatility(weights, annualized_cov_matrix)
    # Avoid division by zero if volatility is zero
    if port_volatility == 0:
        return np.inf
    return -port_return / port_volatility

# Define the constraints (sum of weights is 1, weights are between 0 and 1)
constraints = ({'type': 'eq', 'fun': lambda weights: np.sum(weights) - 1})
bounds = tuple((0, 1) for asset in range(len(daily_returns.columns)))

# Define the initial guess (equal weights)
initial_weights = np.array([1./len(daily_returns.columns)] * len(daily_returns.columns))

# Run the optimizer to find the MSR portfolio weights
optimization_result_msr = minimize(
    negative_sharpe_ratio,
    initial_weights,
    args=(expected_returns, annualized_cov_matrix),
    method='SLSQP',
    bounds=bounds,
    constraints=constraints
)

# Extract the optimal weights for the MSR portfolio
msr_weights = optimization_result_msr.x

# Display the MSR weights
print("\nMaximum Sharpe Ratio Portfolio Weights:")
print(pd.Series(msr_weights, index=daily_returns.columns))
```

## Calculate risk and return for efficient frontier

### Subtask:
Calculate the portfolio return and volatility for a range of portfolio weights to trace the Efficient Frontier.


```python
from scipy.optimize import minimize

def get_efficient_frontier(n_points, expected_returns, annualized_cov_matrix):
    """
    Calculates the efficient frontier for a portfolio.

    Args:
        n_points (int): The number of points to generate on the efficient frontier.
        expected_returns (pd.Series): Annualized expected returns for each asset.
        annualized_cov_matrix (pd.DataFrame): Annualized covariance matrix.

    Returns:
        tuple: A tuple containing two numpy arrays:
               - port_volatilities: Annualized portfolio volatilities for each point on the frontier.
               - port_returns: Annualized portfolio returns (target returns) for each point on the frontier.
    """
    # Initialize arrays to store results
    port_returns = []
    port_volatilities = []

    # Generate a range of target returns
    min_return = expected_returns.min()
    max_return = expected_returns.max()
    target_returns = np.linspace(min_return, max_return, n_points)

    # Define constraints
    constraints = (
        {'type': 'eq', 'fun': lambda weights: np.sum(weights) - 1},  # Sum of weights is 1
        {'type': 'eq', 'fun': lambda weights: portfolio_return(weights, expected_returns / trading_days) - target_return} # Portfolio return equals target return
    )
    bounds = tuple((0, 1) for asset in range(len(expected_returns))) # Weights must be between 0 and 1

    # Define initial guess (equal weights)
    initial_weights = np.array([1./len(expected_returns)] * len(expected_returns))

    # Iterate through target returns and find the minimum volatility portfolio
    for target_return in target_returns:
        # Define the objective function to minimize (portfolio volatility)
        def objective_function(weights):
            return portfolio_volatility(weights, annualized_cov_matrix)

        # Run the optimizer
        optimization_result = minimize(
            objective_function,
            initial_weights,
            method='SLSQP',
            bounds=bounds,
            constraints=constraints
        )

        # Store the results if optimization was successful
        if optimization_result.success:
            port_volatilities.append(optimization_result.fun)
            port_returns.append(target_return)

    return np.array(port_volatilities), np.array(port_returns)

# Calculate the efficient frontier with 50 points
port_volatilities, port_returns = get_efficient_frontier(
    50, expected_returns, annualized_cov_matrix
)

# Display the first few calculated portfolio volatilities and returns
print("First 5 Portfolio Volatilities on the Efficient Frontier:", port_volatilities[:5])
print("First 5 Portfolio Returns on the Efficient Frontier:", port_returns[:5])
```

```python
import streamlit as st
import plotly.graph_objects as go
import numpy as np
import pandas as pd

def run_simulation(n_scenarios, mu, sigma):
    """
    Runs a Monte Carlo simulation of portfolio growth using Geometric Brownian Motion
    and visualizes the results using Plotly with improved hover and PNG export.

    Args:
        n_scenarios (int): The number of simulation scenarios.
        mu (float): The annualized expected return.
        sigma (float): The annualized volatility.
    """
    n_time_periods = portfolio_daily_return.shape[0]
    initial_investment = 1000

    # Generate a random walk
    daily_random_walk = np.random.standard_normal((n_time_periods, n_scenarios))

    # Calculate the daily price changes
    daily_changes = np.exp((mu - 0.5 * sigma**2) * (1/252) + sigma * daily_random_walk * np.sqrt(1/252))

    # Initialize simulated_prices array
    simulated_prices = np.zeros((n_time_periods + 1, n_scenarios))
    simulated_prices[0] = initial_investment

    # Simulate price paths
    for t in range(1, n_time_periods + 1):
        simulated_prices[t] = simulated_prices[t-1] * daily_changes[t-1]

    # Convert to DataFrame for easier handling
    simulated_prices_df = pd.DataFrame(simulated_prices)

    # Create interactive wealth trajectories plot with improved hover
    fig_trajectories = go.Figure()
    
    # Sample a subset of trajectories for visualization performance
    sample_size = min(50, n_scenarios)
    sample_indices = np.random.choice(n_scenarios, sample_size, replace=False)
    
    for i in sample_indices:
        fig_trajectories.add_trace(go.Scatter(
            y=simulated_prices_df.iloc[:, i],
            mode='lines',
            line=dict(width=1, color='rgba(100,149,237,0.3)'),
            showlegend=False,
            # Simplified hover - only essential info
            hovertemplate='T%{x}: $%{y:,.0f}<extra></extra>',
            name=f'Path {i+1}'
        ))
    
    # Add mean trajectory with enhanced styling
    mean_trajectory = simulated_prices_df.mean(axis=1)
    fig_trajectories.add_trace(go.Scatter(
        y=mean_trajectory,
        mode='lines',
        line=dict(width=4, color='#FF4B4B', dash='solid'),
        name='Mean Path',
        # Compact hover for mean
        hovertemplate='<b>Mean</b><br>T%{x}: $%{y:,.0f}<extra></extra>'
    ))
    
    # Add key statistics as floating annotation
    final_mean = mean_trajectory.iloc[-1]
    fig_trajectories.add_annotation(
        x=0.02, y=0.98,
        xref="paper", yref="paper",
        text=f"<b>Quick Stats</b><br>Scenarios: {n_scenarios:,}<br>Final Mean: ${final_mean:,.0f}<br>μ: {mu:.1%} | σ: {sigma:.1%}",
        showarrow=False,
        bgcolor="rgba(255,255,255,0.9)",
        bordercolor="rgba(0,0,0,0.2)",
        borderwidth=1,
        font=dict(size=10, color="black"),
        align="left"
    )
    
    fig_trajectories.update_layout(
        title={
            'text': '🚀 Monte Carlo Portfolio Trajectories',
            'x': 0.5,
            'font': {'size': 18}
        },
        xaxis_title='Time Periods',
        yaxis_title='Portfolio Value ($)',
        template='plotly_white',
        height=600,
        width=1000,
        # Improved hover mode
        hovermode='x unified',
        # Reduce hover distance for less obstruction
        hoverdistance=20
    )

    # Create histogram with compact hover
    terminal_wealth = simulated_prices_df.iloc[-1]
    
    fig_histogram = go.Figure()
    fig_histogram.add_trace(go.Histogram(
        x=terminal_wealth,
        nbinsx=50,
        marker=dict(color='skyblue', line=dict(color='navy', width=1)),
        # Minimal hover info
        hovertemplate='$%{x:,.0f}<br>Count: %{y}<extra></extra>',
        name='Distribution'
    ))
    
    # Add statistical lines with cleaner annotations
    mean_terminal = terminal_wealth.mean()
    median_terminal = terminal_wealth.median()
    percentile_5 = np.percentile(terminal_wealth, 5)
    percentile_95 = np.percentile(terminal_wealth, 95)
    
    # Add vertical lines for key statistics
    for value, color, name in [
        (mean_terminal, "#FF4B4B", "Mean"),
        (median_terminal, "#00CC44", "Median"),
        (percentile_5, "#FF8C00", "5th %"),
        (percentile_95, "#FF8C00", "95th %")
    ]:
        fig_histogram.add_vline(
            x=value, 
            line_dash="dash" if "%" in name else "solid", 
            line_color=color,
            line_width=2,
            annotation_text=f"{name}: ${value:,.0f}",
            annotation_position="top",
            annotation_font_size=9
        )
    
    # Compact stats box
    prob_loss = (terminal_wealth < initial_investment).mean()
    cvar_5 = terminal_wealth[terminal_wealth <= percentile_5].mean()
    
    fig_histogram.add_annotation(
        x=0.98, y=0.95,
        xref="paper", yref="paper",
        text=f"<b>Risk Metrics</b><br>Loss Prob: {prob_loss:.1%}<br>VaR(5%): ${percentile_5:,.0f}<br>CVaR(5%): ${cvar_5:,.0f}",
        showarrow=False,
        bgcolor="rgba(255,255,255,0.9)",
        bordercolor="rgba(0,0,0,0.2)",
        borderwidth=1,
        font=dict(size=10, color="black"),
        align="left",
        xanchor="right"
    )
    
    fig_histogram.update_layout(
        title={
            'text': '📊 Terminal Wealth Distribution',
            'x': 0.5,
            'font': {'size': 18}
        },
        xaxis_title='Terminal Wealth ($)',
        yaxis_title='Frequency',
        template='plotly_white',
        height=500,
        width=900,
        showlegend=False
    )

    # Display key metrics in console
    print(f"📈 Monte Carlo Results ({n_scenarios:,} scenarios):")
    print(f"   Mean Terminal: ${mean_terminal:,.0f}")
    print(f"   Median Terminal: ${median_terminal:,.0f}")
    print(f"   Std Dev: ${terminal_wealth.std():,.0f}")
    print(f"   Loss Probability: {prob_loss:.2%}")
    print(f"   95% VaR: ${percentile_5:,.0f}")
    print(f"   95% CVaR: ${cvar_5:,.0f}")

    # Export plots as PNG (requires kaleido: pip install kaleido)
    try:
        # Save high-quality PNG images
        fig_trajectories.write_image('monte_carlo_trajectories.png', 
                                   width=1000, height=600, scale=2)
        fig_histogram.write_image('monte_carlo_distribution.png', 
                                width=900, height=500, scale=2)
        
        print("\n💾 High-resolution images exported:")
        print("   • monte_carlo_trajectories.png")
        print("   • monte_carlo_distribution.png")
        
    except Exception as e:
        print(f"\n⚠️  PNG export failed: {e}")
        print("   Install kaleido: pip install kaleido")

    # Save interactive HTML versions
    fig_trajectories.write_html('monte_carlo_trajectories.html')
    fig_histogram.write_html('monte_carlo_distribution.html')
    
    # Display plots
    fig_trajectories.show()
    fig_histogram.show()
    
    print("\n📱 Interactive HTML files saved:")
    print("   • monte_carlo_trajectories.html")
    print("   • monte_carlo_distribution.html")
    
    return mean_terminal, median_terminal

# Enhanced dashboard header
print("🎮 Monte Carlo Simulation Dashboard - Enhanced")
print("=" * 60)

# Current parameters display
print(f"📊 Simulation Parameters:")
print(f"   Expected Return (μ): {mu:.2%}")
print(f"   Volatility (σ): {sigma:.2%}")
print(f"   Scenarios: {n_scenarios:,}")
print(f"   Time Periods: {n_time_periods:,}")

# Run enhanced simulation
mean_wealth, median_wealth = run_simulation(n_scenarios, mu, sigma)

# Installation reminder
print("\n" + "="*60)
print("💡 For PNG export capability:")
print("   pip install kaleido")
print("\n🚀 For full interactivity:")
print("   • Streamlit: streamlit run app.py")
print("   • Dash: python dash_app.py")
print("   • Jupyter Lab with widgets")
```

```python
import plotly.graph_objects as go
import pandas as pd
import numpy as np
import yfinance as yf
from IPython import get_ipython

# === TREEMAP INTERACTIVO DEL PORTAFOLIO MSR CON DATOS REALES ===
print("📊 Generando Treemap del Portafolio MSR con datos reales...")

try:
    # 1. EXTRAER VARIABLES REALES DEL KERNEL
    ipython = get_ipython()
    if ipython:
        namespace = ipython.user_ns
    else:
        namespace = globals()
    
    # Obtener variables del portafolio MSR
    msr_weights = namespace.get('msr_weights')
    tickers = namespace.get('tickers')
    expected_returns = namespace.get('expected_returns')
    
    print(f"✅ Variables encontradas:")
    print(f"   - msr_weights: {'Sí' if msr_weights is not None else 'No'}")
    print(f"   - tickers: {'Sí' if tickers is not None else 'No'} - {tickers if tickers else 'N/A'}")
    print(f"   - expected_returns: {'Sí' if expected_returns is not None else 'No'}")
    
    if msr_weights is not None and tickers is not None:
        # 2. FILTRAR ACTIVOS CON PESO SIGNIFICATIVO (>0.5%)
        significant_assets = []
        
        for i, (ticker, weight) in enumerate(zip(tickers, msr_weights)):
            if weight > 0.005:  # Peso > 0.5%
                # Obtener retorno esperado
                if expected_returns is not None:
                    if isinstance(expected_returns, pd.Series):
                        exp_return = expected_returns.iloc[i] if i < len(expected_returns) else 0.10
                    else:
                        exp_return = expected_returns[i] if i < len(expected_returns) else 0.10
                else:
                    exp_return = 0.10
                
                significant_assets.append({
                    'ticker': ticker,
                    'weight': weight,
                    'expected_return': exp_return
                })
        
        print(f"🎯 Activos significativos encontrados: {len(significant_assets)}")
        for asset in significant_assets:
            print(f"   • {asset['ticker']}: {asset['weight']*100:.2f}%")
        
        # 3. OBTENER INFORMACIÓN DE SECTORES USANDO YAHOO FINANCE
        print("\n🔍 Obteniendo información de sectores desde Yahoo Finance...")
        portfolio_data = []
        
        for asset in significant_assets:
            ticker_symbol = asset['ticker']
            print(f"   Procesando: {ticker_symbol}...")
            
            try:
                ticker_obj = yf.Ticker(ticker_symbol)
                info = ticker_obj.info
                
                # Extraer información
                sector = info.get('sector', 'N/A')
                industry = info.get('industry', 'N/A')
                name = info.get('shortName', ticker_symbol)
                
                # Clasificación especial para criptomonedas y bonos
                if sector == 'N/A':
                    if any(crypto in ticker_symbol.upper() for crypto in ['BTC', 'ETH', 'PAXG']):
                        sector = 'Criptoactivos'
                    elif any(bond in ticker_symbol.upper() for bond in ['TLT', 'IEF', 'MBB']):
                        sector = 'Bonos'
                    else:
                        sector = 'Otros'
                
                portfolio_data.append({
                    'ticker': ticker_symbol,
                    'name': name,
                    'sector': sector,
                    'industry': industry,
                    'weight': asset['weight'],
                    'expected_return': asset['expected_return']
                })
                
            except Exception as e:
                print(f"      ⚠️ Error con {ticker_symbol}: {e}")
                # Clasificación manual como fallback
                if any(crypto in ticker_symbol.upper() for crypto in ['BTC', 'ETH', 'PAXG']):
                    sector = 'Criptoactivos'
                elif any(bond in ticker_symbol.upper() for bond in ['TLT', 'IEF', 'MBB']):
                    sector = 'Bonos'
                elif ticker_symbol in ['NVDA', 'GOOG', 'GOOGL', 'AAPL']:
                    sector = 'Tecnología'
                else:
                    sector = 'Otros'
                
                portfolio_data.append({
                    'ticker': ticker_symbol,
                    'name': ticker_symbol,
                    'sector': sector,
                    'industry': 'N/A',
                    'weight': asset['weight'],
                    'expected_return': asset['expected_return']
                })
        
        # 4. CREAR DATAFRAME Y AGRUPAR POR SECTORES
        df = pd.DataFrame(portfolio_data)
        print(f"\n📊 Datos del portafolio procesados:")
        print(df[['ticker', 'name', 'sector', 'weight']].round(4))
        
        # Agrupar por sectores
        sector_summary = df.groupby('sector').agg({
            'weight': 'sum',
            'expected_return': 'mean'
        }).round(4)
        
        print(f"\n📈 Resumen por sectores:")
        print(sector_summary)
        
        # 5. PREPARAR DATOS PARA TREEMAP JERÁRQUICO
        sectors = df['sector'].unique()
        sector_weights = df.groupby('sector')['weight'].sum()
        sector_returns = df.groupby('sector')['expected_return'].mean()
        
        # Estructurar datos para Plotly Treemap
        ids = ['Portafolio MSR'] + sectors.tolist() + df['ticker'].tolist()
        labels = ['Portafolio MSR'] + sectors.tolist() + [f"{row['name']} ({row['ticker']})" for _, row in df.iterrows()]
        parents = [''] + ['Portafolio MSR'] * len(sectors) + df['sector'].tolist()
        values = [df['weight'].sum()] + sector_weights.tolist() + df['weight'].tolist()
        
        # Calcular contribución al rendimiento (peso * retorno esperado)
        total_return = sum(df['weight'] * df['expected_return'])
        contribution = [total_return]  # Portafolio total
        contribution += [sector_weights[sector] * sector_returns[sector] for sector in sectors]  # Sectores
        contribution += [row['weight'] * row['expected_return'] for _, row in df.iterrows()]  # Activos
        
        # 6. SISTEMA DE COLORES BASADO EN CONTRIBUCIÓN (REFINADO)
        colors = []
        for contrib in contribution:
            if contrib > 0.25:  # Contribución excelente (ej. ETH ~40%)
                colors.append('rgba(22, 163, 74, 0.9)')   # Verde muy oscuro
            elif contrib > 0.15:  # Contribución muy alta (ej. BTC ~20%)
                colors.append('rgba(34, 197, 94, 0.8)')   # Verde intenso
            elif contrib > 0.05:  # Contribución alta
                colors.append('rgba(74, 222, 128, 0.8)')  # Verde medio
            elif contrib > 0:  # Contribución positiva (ej. GOOG, GOOGL)
                colors.append('rgba(134, 239, 172, 0.8)') # Verde claro
            elif contrib < -0.1:  # Contribución muy negativa
                colors.append('rgba(239, 68, 68, 0.8)')   # Rojo intenso
            elif contrib < 0:  # Contribución negativa
                colors.append('rgba(252, 165, 165, 0.8)') # Rojo claro
            else:  # Neutral
                colors.append('rgba(156, 163, 175, 0.8)') # Gris
        
        # Color especial para la raíz del portafolio
        colors[0] = 'rgba(107, 114, 128, 0.8)'
        
        # 7. CREAR TREEMAP INTERACTIVO
        fig = go.Figure(go.Treemap(
            ids=ids,
            labels=labels,
            parents=parents,
            values=[v * 100 for v in values],  # Convertir a porcentajes
            text=[f"Contribución: {c*100:.2f}%" for c in contribution],
            textinfo="label+value",
            hovertemplate='<b>%{label}</b><br>Peso: %{value:.1f}%<br>%{text}<extra></extra>',
            pathbar={'visible': True},
            marker={
                'colors': colors,
                'line': {'width': 1, 'color': '#fff'}
            },
            branchvalues='total',
            tiling={'packing': 'squarify'}
        ))
        
        # 8. CONFIGURAR LAYOUT
        fig.update_layout(
            title={
                'text': "📊 Composición del Portafolio MSR - Análisis Detallado",
                'x': 0.5,
                'font': {'size': 16}
            },
            margin={'t': 50, 'l': 10, 'r': 10, 'b': 10},
            autosize=True,
            paper_bgcolor='#FFFFFF',
            plot_bgcolor='#FFFFFF',
            font={'color': '#374151'},
            height=500
        )
        
        # 9. MOSTRAR GRÁFICO
        print("\n🎨 Mostrando Treemap del Portafolio MSR...")
        fig.show()
        
        # 10. MOSTRAR ESTADÍSTICAS FINALES CON CÁLCULOS CORREGIDOS
        print("\n" + "="*60)
        print("📈 RESUMEN FINAL DEL PORTAFOLIO MSR")
        print("="*60)
        print(f"Total de activos significativos: {len(df)}")
        print(f"Peso total analizado: {df['weight'].sum()*100:.1f}%")
        print(f"Rendimiento total del portafolio: {total_return*100:.2f}%")
        print(f"Sectores representados: {len(sectors)}")
        
        print("\nDistribución por sectores:")
        for sector in sectors:
            weight_pct = sector_weights[sector] * 100
            return_pct = sector_returns[sector] * 100
            contrib_pct = sector_weights[sector] * sector_returns[sector] * 100
            print(f"  • {sector}: {weight_pct:.1f}% (retorno esperado: {return_pct:.1f}%, contribución: {contrib_pct:.2f}%)")
        
        print("\nActivos principales (ordenados por peso):")
        for _, row in df.sort_values('weight', ascending=False).iterrows():
            weight_pct = row['weight'] * 100
            return_pct = row['expected_return'] * 100
            contrib_pct = row['weight'] * row['expected_return'] * 100
            contrib_to_total = (contrib_pct / (total_return * 100)) * 100  # % de la contribución total
            print(f"  • {row['ticker']}: {weight_pct:.1f}% (retorno esperado: {return_pct:.1f}%, contribución: {contrib_pct:.2f}% del total)")
        
        print(f"\n🎯 Verificación de cálculos:")
        print(f"   Suma de contribuciones individuales: {sum(df['weight'] * df['expected_return'])*100:.2f}%")
        print(f"   Rendimiento total calculado: {total_return*100:.2f}%")
        
        # Mostrar datos raw para debugging
        print(f"\n🔍 Datos detallados para verificación:")
        for _, row in df.iterrows():
            peso = row['weight']
            retorno = row['expected_return']
            contribucion = peso * retorno
            print(f"   {row['ticker']}: peso={peso:.4f} × retorno={retorno:.4f} = contribución={contribucion:.4f}")
        
        print("="*60)
        
    else:
        print("❌ No se encontraron las variables necesarias del portafolio MSR")
        print("💡 Asegúrate de haber ejecutado las celdas de optimización del portafolio")
        
        # Mostrar variables disponibles
        portfolio_vars = [var for var in namespace.keys() 
                         if any(keyword in var.lower() for keyword in ['msr', 'weight', 'ticker', 'portfolio'])]
        print(f"🔍 Variables relacionadas encontradas: {portfolio_vars[:10]}")

except Exception as e:
    print(f"❌ Error: {e}")
    import traceback
    print("🔧 Detalles del error:")
    print(traceback.format_exc())
```

```python
import json
import pandas as pd
import numpy as np
from IPython import get_ipython
from datetime import datetime

# --- CONFIGURACIÓN ---
output_filename = 'portfolio_analysis_results.json'

def capture_portfolio_analysis():
    """Captura todos los resultados del análisis de portafolio calculados en el notebook."""
    
    # Obtener el kernel actual
    ipython = get_ipython()
    if ipython:
        # Variables del namespace
        namespace = ipython.user_ns
        print("🔍 Capturando resultados del análisis de portafolio...")
    else:
        namespace = globals()
        print("⚠️ Ejecutando fuera de IPython, usando globals()")
    
    # Diccionario para almacenar todos los resultados
    portfolio_results = {
        'timestamp': datetime.now().isoformat(),
        'analysis_type': 'Portfolio Analysis Dashboard',
        'data_summary': {},
        'portfolios': {},
        'monte_carlo': {},
        'efficient_frontier': {},
        'performance_metrics': {},
        'visualizations': {}
    }
    
    try:
        # 1. DATOS BÁSICOS
        if 'tickers' in namespace:
            portfolio_results['data_summary']['tickers'] = namespace['tickers']
            print(f"✅ Tickers: {namespace['tickers']}")
        
        if 'data_cleaned' in namespace:
            df = namespace['data_cleaned']
            portfolio_results['data_summary']['data_shape'] = list(df.shape)
            portfolio_results['data_summary']['date_range'] = {
                'start': str(df.index.min()),
                'end': str(df.index.max())
            }
            portfolio_results['data_summary']['total_days'] = len(df)
            print(f"✅ Datos: {df.shape[0]} días, {df.shape[1]} activos")
        
        # 2. MÉTRICAS DE RENDIMIENTOS
        if 'expected_returns' in namespace:
            returns = namespace['expected_returns']
            portfolio_results['performance_metrics']['expected_returns'] = {
                ticker: float(ret) for ticker, ret in returns.items()
            }
            print(f"✅ Rendimientos esperados capturados para {len(returns)} activos")
        
        # 3. PORTAFOLIOS CALCULADOS
        portfolios_found = 0
        
        # Global Minimum Variance (GMV)
        if 'gmv_weights' in namespace and 'gmv_portfolio_return' in namespace:
            portfolio_results['portfolios']['gmv'] = {
                'name': 'Global Minimum Variance',
                'weights': [float(w) for w in namespace['gmv_weights']],
                'expected_return': float(namespace['gmv_portfolio_return']),
                'volatility': float(namespace['gmv_portfolio_volatility'])
            }
            portfolios_found += 1
        
        # Equal Weight (EW)
        if 'ew_weights' in namespace:
            portfolio_results['portfolios']['equal_weight'] = {
                'name': 'Equal Weight',
                'weights': [float(w) for w in namespace['ew_weights']],
                'expected_return': float(namespace['ew_portfolio_return']),
                'volatility': float(namespace['ew_portfolio_volatility'])
            }
            portfolios_found += 1
        
        # Maximum Sharpe Ratio (MSR)
        if 'msr_weights' in namespace:
            portfolio_results['portfolios']['msr'] = {
                'name': 'Maximum Sharpe Ratio',
                'weights': [float(w) for w in namespace['msr_weights']],
                'expected_return': float(namespace['msr_portfolio_return']),
                'volatility': float(namespace['msr_portfolio_volatility'])
            }
            portfolios_found += 1
        
        print(f"✅ {portfolios_found} portafolios capturados")
        
        # 4. SIMULACIÓN MONTE CARLO
        if 'mean_terminal_wealth' in namespace:
            portfolio_results['monte_carlo'] = {
                'scenarios': int(namespace.get('n_scenarios', 0)),
                'time_periods': int(namespace.get('n_time_periods', 0)),
                'initial_investment': int(namespace.get('initial_investment', 0)),
                'mean_terminal_wealth': float(namespace['mean_terminal_wealth']),
                'median_terminal_wealth': float(namespace['median_terminal_wealth']),
                'wealth_statistics': {
                    'mean': float(namespace.get('mean_wealth', 0)),
                    'median': float(namespace.get('median_wealth', 0))
                }
            }
            print(f"✅ Monte Carlo: {namespace.get('n_scenarios', 0)} escenarios simulados")
        
        # 5. FRONTERA EFICIENTE
        if 'port_returns' in namespace and 'port_volatilities' in namespace:
            returns_array = namespace['port_returns']
            vol_array = namespace['port_volatilities']
            portfolio_results['efficient_frontier'] = {
                'num_portfolios': len(returns_array),
                'return_range': [float(returns_array.min()), float(returns_array.max())],
                'volatility_range': [float(vol_array.min()), float(vol_array.max())],
                'risk_free_rate': float(namespace.get('risk_free_rate', 0))
            }
            print(f"✅ Frontera Eficiente: {len(returns_array)} portafolios generados")
        
        # 6. MÉTRICAS DE RENDIMIENTO ADICIONALES
        if 'max_drawdown' in namespace:
            portfolio_results['performance_metrics']['max_drawdown'] = float(namespace['max_drawdown'])
        
        if 'annualized_return' in namespace:
            portfolio_results['performance_metrics']['annualized_return'] = float(namespace['annualized_return'])
            portfolio_results['performance_metrics']['annualized_volatility'] = float(namespace['annualized_volatility'])
        
        # 7. ARCHIVOS GENERADOS
        generated_files = []
        potential_files = [
            'portfolio_metrics.json',
            'efficient_frontier_interactive.html',
            'monte_carlo_distribution.html',
            'portfolio_growth_interactive.html',
            'efficient_frontier.png',
            'monte_carlo_distribution.png',
            'portfolio_growth.png'
        ]
        
        import os
        for file in potential_files:
            if os.path.exists(file):
                generated_files.append(file)
        
        portfolio_results['visualizations']['generated_files'] = generated_files
        print(f"✅ {len(generated_files)} archivos de visualización encontrados")
        
        # 8. RESUMEN DE VARIABLES EN EL KERNEL
        interesting_vars = [var for var in namespace.keys() 
                          if not var.startswith('_') and 
                          var not in ['In', 'Out', 'exit', 'quit', 'get_ipython']]
        
        portfolio_results['kernel_summary'] = {
            'total_variables': len(interesting_vars),
            'key_variables': [var for var in interesting_vars 
                            if any(keyword in var.lower() for keyword in 
                                 ['portfolio', 'weight', 'return', 'risk', 'sharpe', 'monte', 'efficient'])]
        }
        
        return portfolio_results
        
    except Exception as e:
        print(f"❌ Error capturando datos: {e}")
        return portfolio_results

def print_analysis_summary(results):
    """Imprime un resumen detallado de los resultados capturados."""
    
    print("\n" + "="*60)
    print("📊 RESUMEN DEL ANÁLISIS DE PORTAFOLIO")
    print("="*60)
    
    # Datos básicos
    if 'data_summary' in results and results['data_summary']:
        data = results['data_summary']
        print(f"\n📈 DATOS:")
        if 'tickers' in data:
            print(f"   Activos: {', '.join(data['tickers'])}")
        if 'data_shape' in data:
            print(f"   Observaciones: {data['data_shape'][0]} días")
        if 'date_range' in data:
            print(f"   Período: {data['date_range']['start']} a {data['date_range']['end']}")
    
    # Portafolios
    if 'portfolios' in results and results['portfolios']:
        print(f"\n🎯 PORTAFOLIOS OPTIMIZADOS:")
        for key, portfolio in results['portfolios'].items():
            print(f"   {portfolio['name']}:")
            print(f"     • Rendimiento: {portfolio['expected_return']:.4f}")
            print(f"     • Volatilidad: {portfolio['volatility']:.4f}")
            print(f"     • Ratio Sharpe: {(portfolio['expected_return'] - results.get('efficient_frontier', {}).get('risk_free_rate', 0)) / portfolio['volatility']:.4f}")
    
    # Monte Carlo
    if 'monte_carlo' in results and results['monte_carlo']:
        mc = results['monte_carlo']
        print(f"\n🎲 SIMULACIÓN MONTE CARLO:")
        print(f"   Escenarios: {mc['scenarios']:,}")
        print(f"   Períodos: {mc['time_periods']:,}")
        print(f"   Inversión inicial: ${mc['initial_investment']:,}")
        print(f"   Riqueza terminal promedio: ${mc['mean_terminal_wealth']:,.2f}")
        print(f"   Riqueza terminal mediana: ${mc['median_terminal_wealth']:,.2f}")
    
    # Archivos generados
    if 'visualizations' in results and 'generated_files' in results['visualizations']:
        files = results['visualizations']['generated_files']
        if files:
            print(f"\n📁 ARCHIVOS GENERADOS ({len(files)}):")
            for file in files:
                print(f"   • {file}")
    
    print("\n" + "="*60)

# --- EJECUCIÓN PRINCIPAL ---
try:
    print("🚀 Iniciando captura de resultados del análisis de portafolio...")
    
    # Capturar todos los resultados
    results = capture_portfolio_analysis()
    
    # Guardar en JSON
    with open(output_filename, 'w', encoding='utf-8') as f:
        json.dump(results, f, indent=2, ensure_ascii=False)
    
    print(f"\n✅ Resultados guardados en: '{output_filename}'")
    
    # Mostrar resumen detallado
    print_analysis_summary(results)
    
    print(f"\n💾 Archivo JSON completo guardado para referencia futura")
    
except Exception as e:
    print(f"❌ Error: {e}")
    print("📝 Creando archivo básico de respaldo...")
    
    basic_results = {
        'timestamp': datetime.now().isoformat(),
        'status': 'error',
        'error_message': str(e),
        'note': 'Error al capturar resultados completos'
    }
    
    with open(output_filename, 'w', encoding='utf-8') as f:
        json.dump(basic_results, f, indent=2, ensure_ascii=False)
    
    print(f"⚠️ Archivo de respaldo guardado: '{output_filename}'")
```
